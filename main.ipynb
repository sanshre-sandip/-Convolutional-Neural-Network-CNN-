```python

import torch
from torchvision import datasets, transforms
import matplotlib.pyplot as plt

# Step 1: Convert images into PyTorch tensors
transform = transforms.ToTensor()

# Step 2: Download and load the MNIST dataset
dataset = datasets.MNIST(
    root="data",
    train=True,
    download=True,
    transform=transform,
)

# Step 3: Select one image and its correct answer
image, label = dataset[4]

# Step 4: Inspect the image
print("Image shape:", image.shape)
print("Correct digit:", label)
print("Minimum pixel value:", image.min().item())
print("Maximum pixel value:", image.max().item())

# Step 5: Display the image
plt.imshow(image.squeeze(), cmap="gray")
plt.title(f"Digit: {label}")
plt.axis("off")
plt.show()
```
